```python
import pandas as pd
import numpy as np
import math
import pickle
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier, VotingClassifier
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, classification_report
import warnings
warnings.filterwarnings("ignore")

# Haversine distance (km)
def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2.0)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2.0)**2
    return 2 * R * np.arcsin(np.sqrt(a))
```

```python
# load
df = pd.read_csv("fraudTest.csv") 
df = df.sample(1200, random_state=42)
  # update filename if different
print("Original shape:", df.shape)
df.info(verbose=False)

# Ensure numeric lat/long
for c in ['lat','long','merch_lat','merch_long']:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors='coerce')

# Remove rows with missing critical coords (or impute if you prefer)
coords = [c for c in ['lat','long','merch_lat','merch_long'] if c in df.columns]
if len(coords)>0:
    df = df.dropna(subset=coords)
    print("After dropping missing coords:", df.shape)
```

```python
# convert time and extract hour/day
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])
df['trans_hour'] = df['trans_date_trans_time'].dt.hour
df['trans_day_of_week'] = df['trans_date_trans_time'].dt.dayofweek

# compute time_diff per card (cc_num); fill NA with 0
if 'cc_num' in df.columns:
    df = df.sort_values(['cc_num','trans_date_trans_time'])
    df['time_diff'] = df.groupby('cc_num')['trans_date_trans_time'].diff().dt.total_seconds().fillna(0)
else:
    # fallback: use unix_time differences (not ideal)
    if 'unix_time' in df.columns:
        df = df.sort_values('trans_date_trans_time')
        df['time_diff'] = df['unix_time'].diff().fillna(0)

# distance using haversine
if set(['lat','long','merch_lat','merch_long']).issubset(df.columns):
    df['distance_km'] = haversine(df['lat'].values,
                                 df['long'].values,
                                 df['merch_lat'].values,
                                 df['merch_long'].values)
else:
    df['distance_km'] = 0.0

# drop original datetime if not needed
df = df.drop(columns=['trans_date_trans_time'], errors='ignore')

df[['trans_hour','trans_day_of_week','time_diff','distance_km']].head()
```

```python
# Drop columns we don't want (IDs / PII / text). adjust as needed.
drop_cols = ['Unnamed: 0','trans_num','first','last','street','merchant','job','dob']
for c in drop_cols:
    if c in df.columns:
        df = df.drop(columns=c)

# If you want to keep cc_num or zip, be aware they are identifiers; drop if not needed
if 'cc_num' in df.columns:
    df = df.drop(columns=['cc_num'])   # usually drop cc_num to avoid leakage

# Choose categorical columns to one-hot encode
categorical_cols = [c for c in ['category','gender','state','city'] if c in df.columns]
print("Categorical columns to encode:", categorical_cols)

# One-hot encode (safe, reproducible order)
X = pd.get_dummies(df.drop(columns=['is_fraud']), columns=categorical_cols, drop_first=True)
y = df['is_fraud'].astype(int)

# Final check
print("Final feature shape:", X.shape)
print("Any non-numeric dtypes:", X.select_dtypes(include=['object','category']).columns.tolist())
print("Missing values in X:", X.isna().sum().sum())
X = X.fillna(0)
```

```python
# split
X_train_df, X_test_df, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# scaler (fit on train)
numeric_cols = X.select_dtypes(include=['int64','float64']).columns.tolist()
scaler = StandardScaler()
X_train = X_train_df.copy()
X_test = X_test_df.copy()

# scale numeric columns in place using DataFrame indices (keeps column structure)
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])



print("Shapes ->", X_train.shape, X_test.shape, "y distribution:", y_train.value_counts().to_dict())
```

```python
# Optional: SMOTE (uncomment to use — you must pip install imbalanced-learn)
use_smote = False
if use_smote:
    try:
        from imblearn.over_sampling import SMOTE
        sm = SMOTE(random_state=42)
        X_res, y_res = sm.fit_resample(X_train, y_train)
        X_train = pd.DataFrame(X_res, columns=X_train.columns)
        y_train = pd.Series(y_res)
        print("After SMOTE:", X_train.shape, y_train.value_counts().to_dict())
    except Exception as e:
        print("SMOTE not available - skip. Error:", e)
```

```python
print("\n🔹 Training Logistic Regression Model...")
lr_clf = LogisticRegression(random_state=42, solver='liblinear', class_weight='balanced')
lr_clf.fit(X_train, y_train)
```

```python
# ============================
#   CELL G – FAST ENSEMBLE MODELS
# ============================

from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score
import numpy as np
import pandas as pd

models = {
    'RandomForest': RandomForestClassifier(
        n_estimators=120,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    ),

    'AdaBoost': AdaBoostClassifier(
        n_estimators=80,
        random_state=42
    )
}

# Try importing XGBoost
try:
    from xgboost import XGBClassifier
    models['XGBoost'] = XGBClassifier(
        n_estimators=100,         # reduced for speed
        learning_rate=0.1,
        max_depth=4,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric='logloss',
        random_state=42,
        n_jobs=-1
    )
    print("XGBoost loaded successfully.")
except Exception as e:
    print("XGBoost NOT available → running without it.")
    print(e)

results = []
fitted = {}

# Train each model
for name, model in models.items():
    print("\n🚀 Training:", name)
    model.fit(X_train, y_train)
    fitted[name] = model

    # Predictions
    y_pred = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:, 1]
    else:
        try:
            y_proba = model.decision_function(X_test)
        except:
            y_proba = np.zeros_like(y_pred, dtype=float)

    # Metrics
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec  = recall_score(y_test, y_pred, zero_division=0)
    f1   = f1_score(y_test, y_pred, zero_division=0)

    try:
        roc = roc_auc_score(y_test, y_proba)
    except:
        roc = 0.0

    results.append({
        'model': name,
        'precision': prec,
        'recall': rec,
        'f1': f1,
        'roc_auc': roc
    })

    print(f"🔹 {name} → Precision: {prec:.4f}  Recall: {rec:.4f}  F1: {f1:.4f}  ROC-AUC: {roc:.4f}")

# Show results
results_df = pd.DataFrame(results).sort_values('roc_auc', ascending=False).reset_index(drop=True)
display(results_df)
```

```python
# =============================
#   CELL H – SELECT & SAVE BEST MODEL
# =============================

# Pick best model based on ROC-AUC
best_model_name = results_df.iloc[0]["model"]
best_model = fitted[best_model_name]

print(f"\n🎉 Best Model Selected: {best_model_name}")

# Save best model
import pickle

with open("fraud_model_ensemble.pkl", "wb") as f:
    pickle.dump(best_model, f)

with open("scaler.pkl", "wb") as f:
    pickle.dump(scaler, f)

print("\n💾 Saved fraud_model_ensemble.pkl and scaler.pkl successfully!")

# Show results table again
results_df
```
